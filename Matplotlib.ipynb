Data Visulation in Python:
Data Visulation in python means representing data in the form of graph ,charts and plots so that we can understand patterns and comparisons easily.


Matplotlib: Matplotlib is a python library used to create :
1:Line graphs
2:Baar Charts
3:Histograms
4:Pie Charts
5:Scatter plots 
6:Subplots
7:And many other visulation

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
x=[1,2,3,4]
y=[10,20,30,40]
plt.plot(x,y)
plt.show()
```

```python
x=np.linspace(0,5,11)
print(x)
y=x**2
print(y)
plt.plot(x,y)
```

```python
x=[1,2,3,4,5]
y=[10,20,15,30,25]
plt.plot(x,y)
#ADDING TITLE:
plt.title("Hello World")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
#Grid: Grid makes the graph easier to read.
x=[1,2,3,4,5]
y=[10,20,15,30,25]
plt.plot(x,y)
plt.title("Student Performance")
plt.xlabel("X-axis")
plt.ylabel("Y-Label")
plt.grid()
plt.show()
```

```python
#Markers: We can show individual data points using markers.
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]
plt.plot(x,y,marker='o')
plt.show()
#NOTE: 'o; means a circular marker.
```

```python
#Bar Chart:A bar chart is useful for comparing different categories.

subjects = ["Maths", "Python", "C++", "DSA"]
marks = [85, 90, 78, 88]
plt.bar(subjects,marks)
plt.title("Subject Marks")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.show()
```

```python
# Horizontal Bar Chart
subjects = ["Maths", "Python", "C++", "DSA"]
marks = [85, 90, 78, 88]
plt.barh(subjects, marks)

plt.title("Subject Marks")
plt.xlabel("Marks")
plt.ylabel("Subjects")

plt.show()
```

```python
# Scatter Plot:  A scatter plot represents individual points.

# Used to understand the relationship between two variables.

hours = [1, 2, 3, 4, 5, 6]
marks = [40, 45, 50, 60, 70, 80]
plt.scatter(hours,marks)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
# Here:

# X → Study Hours
# Y → Marks

# We can visually observe whether marks increase as study hours increase.
```

```python
# Histogram:  A histogram is used to show the distribution of numerical data.

marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75, 78, 80, 85, 90]

plt.hist(marks)

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
```

```python
# Pie Chart:  A pie chart shows how a whole is divided into different parts.
subjects = ["Maths", "Python", "C++", "DSA"]
marks = [85, 90, 78, 88]

plt.pie(marks, labels=subjects)

plt.title("Marks Distribution")

plt.show()
```

```python
# Legend: When we have multiple lines/data sets, we can use a legend.
x = [1, 2, 3, 4, 5]

maths = [60, 65, 70, 75, 80]
python = [70, 75, 80, 85, 90]

plt.plot(x, maths, label="Maths")
plt.plot(x, python, label="Python")

plt.legend()

plt.show()
```

```python
# Multiple Plots: We can plot multiple datasets on the same graph.
x = [1, 2, 3, 4, 5]

y1 = [10, 20, 30, 40, 50]
y2 = [5, 15, 25, 35, 45]

plt.plot(x, y1, label="Data 1")
plt.plot(x, y2, label="Data 2")

plt.legend()

plt.show()
```
